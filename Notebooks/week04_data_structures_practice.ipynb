## 1. 리스트 만들기 · 인덱싱 · 슬라이싱
대괄호 `[ ]`에 쉼표로 나열하면 **리스트**입니다. 순서가 있어서 번호(**인덱스**, 0부터!)로 꺼낼 수 있습니다.

````python
days = ['월', '화', '수', '목', '금', '토', '일']
days
````

````python
days[0], days[1]   # 첫 번째, 두 번째 — 인덱스는 0부터!
````

````python
days[-1]   # 마지막 원소 → '일' (음수 인덱스는 뒤에서부터)
````

````python
days[1:4]   # 슬라이싱 — 화, 수, 목 (끝 인덱스 4는 미포함)
````

````python
days[:3], days[3:]   # 처음 3개 / 네 번째부터 끝까지
````

````python
len(days)   # 원소 개수 → 7
````

> ⚠ 범위를 벗어난 인덱스는 `IndexError` — `days[7]`은 '여덟 번째'를 요구하는 것이므로 오류입니다. 반면 슬라이스는 범위를 넘어도 오류 대신 알아서 잘라냅니다.

## 2. 리스트 메서드 — append · sort
리스트는 자기만의 기능(**메서드**)을 갖고 있습니다. `.`(점)을 찍어 부릅니다.

````python
todo = ["수업 듣기", "과제 하기"]
todo.append("운동 하기")   # 맨 뒤에 추가
todo
````

````python
nums = [64, 25, 90, 12]
nums.sort()   # 오름차순 정렬 — 원본 자체가 바뀜
nums          # [12, 25, 64, 90]
````

````python
nums.sort(reverse=True)   # 내림차순
nums                      # [90, 64, 25, 12]
````

````python
sorted(nums)   # 정렬된 '새 리스트'를 돌려줌 → [12, 25, 64, 90]
nums           # 원본은 그대로 [90, 64, 25, 12]
````

> ⚠ `.sort()`는 원본을 바꾸고 `None`을 돌려줍니다 — `nums = nums.sort()`처럼 담으면 리스트가 사라집니다! 새 리스트가 필요하면 `sorted(nums)`를 쓰세요.

## 3. 리스트 계산 — 일주일 기온
이번 실습의 주인공 데이터입니다. **이 숫자들은 5주차 넘파이 예제에 그대로 다시 등장합니다.**

````python
high = [18.5, 19.2, 21.0, 19.8, 22.3, 21.5, 23.4]   # 최고 기온
low  = [ 9.2, 10.1, 12.0, 11.4, 13.1, 12.8, 14.0]   # 최저 기온
days = ['월', '화', '수', '목', '금', '토', '일']

len(high), len(low), len(days)   # 세 리스트 길이가 같다 = 7일치 데이터
````

````python
sum(high) / len(high)   # 주간 평균 최고기온 = 합계 ÷ 개수 → 20.814285714285713
````

````python
max(high), min(low)   # 주간 최고 23.4℃ / 주간 최저 9.2℃
````

````python
high.index(max(high))        # 최고 기온이 몇 번째(인덱스)? → 6
days[high.index(max(high))]  # 그 요일 → '일'
````

> 💡 **감 잡기**: `sum` · `len` · `max` · `min`은 오늘 쓴 원시 도구입니다. 다음 주 넘파이에서는 `high.mean()`, `high.max()`처럼 배열이 스스로 계산해 줍니다.

## 4. 튜플과 언패킹
**튜플 `( )`** 은 괄호만 다른 '못 바꾸는 리스트'입니다. 한 번 만들면 수정이 불가능하고, 바뀌면 안 되는 데이터(좌표 등)에 씁니다.

````python
point = (3, 4)         # (x, y) 좌표 — 바뀌면 안 되는 값
point[0], point[1]     # 튜플도 인덱스로 읽기 → (3, 4)
````

````python
x, y = point   # 언패킹 — 튜플의 값을 여러 변수에 한 번에 담기
x + y          # 7
````

````python
a, b = 1, 2
a, b = b, a    # 두 변수의 값 교환 — 파이썬의 명물 문법
a, b           # (2, 1)
````

````python
# point[0] = 99   # TypeError — 튜플은 값을 바꿀 수 없다!
point
````

> ⚠ 언패킹은 **개수가 맞아야** 합니다 — `x, y = (1, 2, 3)`은 `ValueError: too many values to unpack`.

## 5. 딕셔너리 — 성적표
**딕셔너리 `{ }`** 는 번호가 아니라 **이름(키)** 으로 값을 찾는 자료구조입니다. '국어 → 88' 같은 짝(쌍)의 모음 — 성적표·전화번호부에 딱입니다.

````python
scores = {"국어": 88, "영어": 92, "수학": 95}   # {키: 값, 키: 값, ...}
scores
````

````python
scores["수학"]   # 키로 값 읽기 → 95
````

````python
scores["과학"] = 90   # 없는 키에 대입 → 새 쌍 추가
scores
````

````python
scores["영어"] = 95   # 있는 키에 대입 → 값 수정
scores
````

````python
scores.keys(), scores.values()   # 키만 / 값만
````

````python
"음악" in scores    # 키가 있는지 확인 → False
scores.get("음악")  # get은 없어도 오류 대신 None을 돌려줌
````

> ⚠ 없는 키를 대괄호로 읽으면 `KeyError`입니다. 확실하지 않으면 `scores.get(키)`가 안전합니다.

## 6. 딕셔너리 순회 — items
`for`와 `items()`를 쓰면 '키-값' 쌍을 하나씩 꺼내며 순회할 수 있습니다.

````python
for subject, score in scores.items():   # 키, 값이 한 쌍으로 나옴
    print(f"{subject}: {score}점")
````

````python
total = 0
for subject, score in scores.items():
    total += score

print("과목 수:", len(scores))                          # 4
print("합계:", total, "/ 평균:", total / len(scores))   # 합계 368 / 평균 92.0
````

## 7. 집합 — 중복 제거
**집합**은 중복을 자동으로 없애고 순서가 없습니다. '방문한 도시에서 중복 빼기' 같은 일에 딱입니다.

````python
visited = ["서울", "부산", "서울", "제주", "부산", "대구", "서울"]
unique = set(visited)   # 리스트 → 집합: 중복이 증발
unique                  # 어떤 순서로 출력될지는 보장되지 않음
````

````python
len(unique)   # 중복 제거 후 개수 → 4
````

````python
"제주" in unique   # 포함 확인 → True (집합의 in은 리스트보다 훨씬 빠름)
````

````python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
a & b, a | b   # 교집합 / 합집합 → ({3, 4}, {1, 2, 3, 4, 5, 6})
````

> 💡 중괄호 `{ }`의 두 얼굴 — 안에 `키: 값`이 있으면 **딕셔너리**, 값만 나열하면 **집합**. 빈 `{ }`는 딕셔너리라서, 빈 집합은 `set()`으로 만듭니다.

## 8. 반복문과 enumerate — 가장 더운 요일
`enumerate(리스트)`는 **인덱스와 값**을 한 쌍씩 줍니다 — '몇 번째인지'가 필요한 순회에 필수입니다.

````python
for i, day in enumerate(days):   # (인덱스, 값)
    print(i, day)                # 월요일은 0번
````

````python
hottest = 0                        # 일단 0번째(월)가 최고라고 가정

for i in range(1, len(high)):      # 나머지 날들을 훑으며
    if high[i] > high[hottest]:    # 더 더운 날을 만나면
        hottest = i                # 최고 자리를 갱신

print(f"가장 더운 날: {days[hottest]} {high[hottest]}℃")   # 일 23.4℃
````

````python
high.index(max(high))   # 지난 절의 '한 줄' 방법 — 결과는 같다 (6)
````

> 💡 **감 잡기**: 8-2의 '첫 값을 최고로 가정 → 더 크면 갱신' 패턴이 `max`의 내부 원리입니다. 다음 주 넘파이의 `high.argmax()`가 이 반복문을 한 단어로 줄여 줍니다.

## 9. 종합 예제 — 주간 기온 리포트
오늘 배운 전부를 사용합니다: **리스트 + 반복문 + f-string + 누적**.

````python
print("=== 한 주 기온 리포트 ===")
for i in range(len(days)):
    daily_range = high[i] - low[i]   # 일교차
    print(f"{days[i]}: 최고 {high[i]}℃ / 최저 {low[i]}℃ (일교차 {daily_range:.1f}℃)")
````

````python
avg_high = sum(high) / len(high)   # 주간 평균
avg_low  = sum(low)  / len(low)

print("--- 주간 요약 ---")
print(f"평균 최고 {avg_high:.1f}℃ / 평균 최저 {avg_low:.1f}℃")
print(f"최고 {max(high)}℃({days[high.index(max(high))]}) / 최저 {min(low)}℃({days[low.index(min(low))]})")
````

> 💡 **감 잡기(선택)**: `high`, `low`의 숫자를 본인 도시의 최근 일주일 기온으로 바꿔 다시 실행해 보세요. 코드는 그대로 둔 채 **데이터만** 바꿔도 결과가 다시 계산됩니다 — 코드와 데이터의 분리가 분석의 핵심입니다.

## 10. 5주차 예고 — 이 리스트가 `np.array`가 되면
오늘 반복문으로 하던 계산이 다음 주에는 **한 줄**이 됩니다. (지금은 실행할 필요 없습니다 — 미리 보기일 뿐!)

```python
import numpy as np

high = np.array([18.5, 19.2, 21.0, 19.8, 22.3, 21.5, 23.4])   # 리스트 → 배열
low  = np.array([ 9.2, 10.1, 12.0, 11.4, 13.1, 12.8, 14.0])

high - low                # 일교차 7개 — 반복문 없이 한 번에!
high.mean(), low.mean()   # 평균도 메서드 하나
days[high.argmax()]       # 가장 더운 요일도 한 줄
```

리스트 → 배열(넘파이) → 표(판다스)의 순서로, 오늘 배운 인덱싱·슬라이싱·순회가 그대로 확장됩니다.

## 오늘의 정리
1. **리스트**: 순서가 있는 묶음 — 인덱스(0부터) · 슬라이싱(끝 미포함) · `append` · `sort`(원본 변형!) vs `sorted`
2. **계산**: `len` · `sum` · `max` · `min` + `.index()`로 위치 찾기 — 평균은 `sum/len`
3. **튜플**은 못 바꾸는 리스트 — 언패킹 `x, y = point`, 교환 `a, b = b, a`
4. **딕셔너리**는 키로 찾는 성적표 — `items()`로 순회, 없는 키는 `get`으로 안전하게
5. **집합**은 중복 없는 묶음 — `set(리스트)`로 중복 제거, `in` · `&` · `|`
