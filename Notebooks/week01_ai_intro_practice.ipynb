## 1. 첫 코드 — print
`print`는 괄호 안의 내용을 화면에 보여주는 명령입니다. 따옴표로 감싼 글자는 **문자열**, 따옴표 없이 쓴 것은 **숫자나 계산식**으로 해석됩니다.

```python
print("2026 컴퓨터코딩 · 데이터 분석 입문")   # 본인 이름이나 학과로 바꿔 보세요
```

```python
print('작은따옴표도 됩니다')    # '...' 도 문자열
print("큰따옴표도 됩니다")      # "..." 도 문자열
```

```python
print(100 + 50)     # 따옴표가 없으면 계산식 → 150
print("100 + 50")   # 따옴표가 있으면 그냥 글자 → 100 + 50
```

```python
print("올해는", 2026, "년")   # 쉼표로 여러 값을 나란히 출력 → 올해는 2026 년
```

> 💡 코드에서 `#` 뒤에 붙은 글자는 **주석**입니다 — 파이썬이 무시하는 메모란. 노트북 곳곳의 `#`은 모두 설명을 위한 메모이며, 실행 결과에는 나타나지 않습니다.

## 2. 계산기로 파이썬
파이썬은 세상에서 가장 비싼(?) 계산기이기도 합니다. 기호만 바꿔 가며 계산해 봅시다.

| 기호 | 의미 |
|---|---|
| `+` `-` `*` | 더하기 · 빼기 · 곱하기 |
| `/` | 나누기 (소수점까지 계산) |
| `//` | 몫 (나누고 정수 부분만) |
| `%` | 나머지 |
| `**` | 거듭제곱 |

```python
365 * 24          # 1년은 몇 시간? → 8760
```

```python
10 / 4            # 나누기는 소수점까지 → 2.5
```

```python
10 // 4, 10 % 4   # 몫과 나머지 → (2, 2)
```

```python
2 ** 10           # 2의 10제곱 → 1024
```

```python
60 * 60 * 24      # 하루는 몇 초? → 86400
```

> ⚠ 나누기 `/`의 결과는 정수로 나누어 떨어져도 항상 실수입니다 — `10 / 2`는 `5`가 아니라 `5.0`.

## 3. 데이터 맛보기 — 평균 구하기
인공지능(AI)의 출발점은 **데이터**입니다. 지난 일주일 기온이라는 작은 데이터로 평균을 구해 봅시다. 대괄호 `[ ]`로 묶은 **리스트**는 여러 값을 한 덩어리로 담는 그릇입니다 (자세한 것은 4주차에).

```python
temps = [17.1, 18.5, 21.3, 20.8, 19.6]   # 최근 5일의 최고 기온
temps
```

```python
sum(temps)   # 모두 더하기 → 97.30000000000001 (끝자리 오차는 컴퓨터가 소수를 다루는 방식 때문 — 정상!)
```

```python
len(temps)   # 데이터가 몇 개인지 세기 → 5
```

```python
sum(temps) / len(temps)   # 평균 = 합계 ÷ 개수 → 19.46
```

```python
max(temps), min(temps)   # 가장 높았던 날과 가장 낮았던 날 → (21.3, 17.1)
```

> 💡 **감 잡기**: "내일 기온은 얼마?"를 맞혀 보라고 하면, 제일 무난한 답은 **평균**입니다. AI도 결국 데이터에서 이런 숫자(패턴)를 찾아 **예측**합니다 — 아래 셀은 오늘 배운 것만으로 만든 '인류 최초의 예측 모델'입니다.

```python
예측 = sum(temps) / len(temps)   # 가장 단순한 예측법: '평균'
실제 = 19.8                      # 실제로 측정된 다음 날 기온

print("내 예측:", 예측)          # 데이터가 알려 준 답
print("실제 기온:", 실제)
print("오차:", round(실제 - 예측, 2))   # 예측과 실제의 차이 → 0.34
```
