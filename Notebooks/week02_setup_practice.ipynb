## 1. 내 파이썬 버전
버전 확인은 트러블슈팅의 첫걸음입니다. "안 돼요"가 아니라 **"파이썬 3.12에서 이런 오류가 났어요"** 라고 말할 수 있게 만들어 줍니다.

```python
import sys   # 파이썬 자신에 대한 정보를 담은 표준 모듈

sys.version   # 버전과 빌드 정보 문자열
```

```python
import platform

platform.python_version()   # '3.12.13' 처럼 깔끔한 버전 번호만
```

```python
sys.executable   # 지금 돌아가는 파이썬 프로그램의 전체 위치
```

```python
if "google.colab" in sys.modules:   # Colab이 몰래 심어 둔 모듈이 있으면
    print("지금은 Google Colab에서 실행 중")     # → Colab
else:
    print("지금은 내 컴퓨터(로컬)에서 실행 중")  # → 그 외
# (조건문 if는 3주차에 본격 배웁니다 — 지금은 '어디서 돌아가는지 알려주는 도구'로만)
```

> 💡 `sys.version`은 정보 덩어리 문자열, `platform.python_version()`은 `'3.12.13'`처럼 깔끔한 번호 — **질문 글이나 보고서에는 후자**가 깔끔합니다. 버전이 3.10 이상이면 이번 학기 내용은 모두 문제없습니다.

## 2. 수학 도구 — math
`import math` 한 줄이면 고급 계산 도구가 생깁니다. `math.` 까지 입력하고 **Tab**을 누르면 쓸 수 있는 기능 목록이 뜹니다(Colab·VS Code 공통).

```python
import math

math.sqrt(2)      # 제곱근 √2 → 1.4142135623730951
```

```python
math.pi           # 원주율 → 3.141592653589793
```

```python
math.floor(3.7), math.ceil(3.2)   # 내림 / 올림 → (3, 4)
```

```python
round(3.14159, 2)   # 반올림(자릿수 지정) — round는 기본 제공이라 import 불필요 → 3.14
```

## 3. 날짜·시간 — datetime
데이터 분석의 단골 손님은 **날짜**입니다. 지금 몇 시인지, 100일 뒤가 며칠인지 물어봅시다.

```python
from datetime import datetime   # datetime 모듈에서 datetime 도구를 꺼내 쓴다

now = datetime.now()   # 지금 이 순간
now
```

```python
now.year, now.month, now.day   # 연, 월, 일 (실행한 순간의 날짜)
```

```python
now.strftime("%Y년 %m월 %d일 %H시 %M분")   # 원하는 모양으로 조립
```

```python
from datetime import timedelta

due = now + timedelta(days=100)   # 100일 뒤 — 과제 마감일 계산에 유용
due.strftime("%Y년 %m월 %d일")
```

> ⚠ `import datetime`과 `from datetime import datetime`은 다릅니다 — 후자는 모듈 안의 **같은 이름의 도구**를 꺼내 쓰는 방법. `from` 형태로 꺼내 쓰면 `datetime.now()`처럼 짧게 쓸 수 있습니다.

## 4. 랜덤 맛보기 — random
무작위 수는 게임·시뮬레이션·샘플 뽑기의 재료입니다. 5주차에서 예제 데이터를 만들 때도 다시 만납니다.

```python
import random

random.seed(42)   # 시드 고정 → 언제 누가 실행해도 같은 결과(재현 가능)
random.random()   # 0 이상 1 미만 실수 → 0.6394267984578837 (시드 42면 전원 동일)
```

```python
random.randint(1, 6)   # 주사위 — 1부터 6까지 정수 하나 (실행마다 다름)
```

```python
menu = ["김밥", "라면", "국밥", "샌드위치", "비빔밥"]
random.choice(menu)   # 오늘의 점심 — 목록에서 하나 뽑기
```

```python
nums = [1, 2, 3, 4, 5]
random.shuffle(nums)   # 순서 섞기 (원본 자체가 바뀜)
nums
```

```python
random.sample(nums, 2)   # 중복 없이 2개 뽑기 — 발표 순서 정하기
```

> ⚠ **시드(seed)** 를 고정하면 난수가 일정한 순서로 나와 결과를 **재현**할 수 있습니다 — 과제와 시뮬레이션의 필수 장치. 시드가 없으면 실행할 때마다 결과가 달라집니다.
