익숙한 파이썬 **리스트**를 넘파이 **배열(ndarray)**로 바꿔 봅시다. 셀의 마지막 줄에 변수명만 쓰면 내용이 표시됩니다.

## 2. 자주 쓰는 배열 생성 함수
| 함수 | 하는 일 |
|---|---|
| `np.arange(시작, 끝, 간격)` | 시작부터 **끝 전까지** 간격씩 |
| `np.zeros(모양)` / `np.ones(모양)` | 0 / 1로 채운 배열 |
| `np.full(모양, 값)` | 원하는 값으로 채운 배열 |
| `np.linspace(시작, 끝, 개수)` | 시작~**끝 포함** 균등 분할 |

⚠ 암기 포인트: `arange`의 끝은 **미포함**, `linspace`의 끝은 **포함**.

```python
np.arange(1, 10, 2)     # [1 3 5 7 9]
```

```python
np.zeros((2, 3))
```

```python
np.ones((2, 2))
```

```python
np.full((2, 2), 7)
```

```python
np.linspace(0, 1, 5)   # [0. 0.25 0.5 0.75 1.]
```

## 5. 인덱싱과 슬라이싱 (1차원)
규칙은 리스트와 같습니다 — 0부터 시작, `a[시작:끝]`에서 끝은 미포함.

```python
a = np.arange(1, 11)
a
```

```python
a[0], a[-1]     # 첫 원소, 마지막 원소
```

```python
a[2:5]          # [3 4 5]
```

```python
a[::2]          # 두 칸씩: [1 3 5 7 9]
```

```python
a[::-1]         # 거꾸로
```

### ⚠ 슬라이스는 '복사'가 아니라 '뷰(view)'
리스트의 슬라이스는 사본이지만, **넘파이의 슬라이스는 원본을 함께 보는 창**입니다. 진짜 사본이 필요하면 `.copy()`를 붙입니다.

```python
b = a[2:5]
b[0] = 99       # b를 바꾸면…
a               # 원본 a도 함께 바뀜!
```

```python
c = a[2:5].copy()   # 진짜 사본
c[0] = 0
a                    # 이번엔 안 바뀜
```

## 6. 2차원 인덱싱 — 행, 열을 쉼표로
`m[행, 열]` — 콤마 앞이 행(가로줄), 뒤가 열(세로줄). `:`는 '그 방향 전체'.

```python
m = np.arange(12).reshape(3, 4)
m
```

```python
m[1, 2]     # 2행 3열 (0부터 셈) → 6
```

```python
m[0]        # 첫 번째 행
```

```python
m[:, 1]     # 두 번째 열 → [1 5 9]
```

```python
m[:2, 1:3]  # 행 0~1, 열 1~2
```

## 7. 조건으로 골라내기 — 불리언 인덱싱
데이터 분석에서 **가장 자주 쓰는** 선택 방법입니다.

```python
score = np.array([78, 92, 65, 88, 54, 95, 71])
score >= 80              # 조건 → True/False 마스크
```

```python
score[score >= 80]       # True 자리만 선택 → [92 88 95]
```

```python
score[(score >= 70) & (score < 90)]   # & (그리고)로 조합
```

> ⚠ 조건 조합은 `and` / `or`가 아니라 **`&` / `|`** 이며, 조건마다 **괄호**로 묶어야 합니다.

## 8. 팬시 인덱싱 — 뽑고 싶은 번호를 나열
리스트로 인덱스를 나열하면 원하는 순서대로(중복·역순 포함) 뽑을 수 있습니다. 결과는 항상 **새 배열(복사)**입니다.

```python
a = np.arange(10) * 10
a[[3, 1, 7]]     # [30 10 70]
```

```python
m[[2, 0, 1]]     # 행 순서 바꾸기
```

## 9. 벡터화 연산 — 반복문 없이 한 번에
리스트와 배열의 결정적 차이를 눈으로 확인해 봅시다.

```python
[1, 2, 3] * 3                      # 리스트: 반복!
```

## 10. 브로드캐스팅 — 모양이 달라도 계산
작은 쪽 모양을 큰 쪽에 맞춰 **자동 복제** 후 계산합니다. 규칙은 **끝자리 모양끼리** 비교해 같거나 한쪽이 1이면 확장.

```python
a = np.arange(4)
a + 10      # 스칼라 10이 모든 원소로 퍼짐
```

```python
m + np.arange(4)    # (3,4) + (4,) → (4,)가 각 행으로 복사
```

```python
np.arange(3).reshape(3, 1) + np.arange(4)   # (3,1)+(1,4) → (3,4)
```

```python
try:
    np.arange(3) + np.arange(4)   # (3,) + (4,) → 불가!
except ValueError as e:
    print("ValueError:", e)
```

## 11. 집계 함수와 `axis`
`sum` · `mean` · `std` · `max` · `min` 모두 같은 규칙.
- `axis=0` : **아래 방향**으로 훑어 같은 **열**끼리 묶기
- `axis=1` : **옆 방향**으로 훍어 같은 **행**끼리 묶기
- 생략하면 전체를 하나로 묶음

헷갈리면 **결과 개수**로 확인 — 3×4 배열에서 열별이면 4개, 행별이면 3개.

```python
m.sum()          # 전체 합 → 66
```

```python
m.sum(axis=0)    # 열별 합 → [12 15 18 21]
```

```python
m.sum(axis=1)    # 행별 합 → [ 6 22 38]
```

```python
m.mean(axis=1)   # 행별 평균
```

```python
m.std(), m.max(), m.min()   # 표준편차, 최댓값, 최솟값
```

## 12. 모양 바꾸기 — `reshape`과 전치 `.T`
원소 수(`size`)가 같아야만 바꿀 수 있습니다. `-1`을 쓰면 그 자리는 자동 계산.

```python
a = np.arange(12)
a.reshape(3, 4)
```

```python
a.reshape(2, -1).shape    # (2, 6) — -1은 '남은 만큼 알아서'
```

```python
a.reshape(3, 4).T         # 전치: 행↔열 맞바꿈
```

## 13. 배열 붙이기 — `vstack` · `hstack`
`v`는 vertical(세로), `h`는 horizontal(가로). 묶을 배열들은 **리스트로** 넘깁니다.

```python
p = np.arange(6).reshape(2, 3)
q = np.full((2, 3), 7)
p, q
```

```python
np.vstack([p, q])    # 아래로 이어붙이기 → (4, 3)
```

```python
np.hstack([p, q])    # 옆으로 이어붙이기 → (2, 6)
```

## 14. 정렬 — `sort`와 `argsort`
- `np.sort(x)` : 값 자체를 정렬
- `np.argsort(x)` : 정렬된 **순서의 인덱스** — '누가 몇 등인지'

`argsort` + 팬시 인덱싱은 **'점수 순서대로 다른 배열을 줄 세우는'** 데이터 분석 단골 패턴입니다.

```python
name  = np.array(['김', '이', '박', '최', '정'])
score = np.array([88, 72, 95, 61, 84])
np.sort(score)            # [61 72 84 88 95]
```

```python
np.argsort(score)         # [3 1 4 0 2] — 최→이→정→김→박 순서
```

```python
name[np.argsort(score)]   # 낮은 점수순으로 이름 배열
```

```python
np.argsort(-score)        # 내림차순(높은 순서) 인덱스
```

## 15. 종합 예제 — 한 주의 기온 데이터
오늘 배운 것 전부를 사용합니다: **만들기 → 계산 → 선택 → 정렬**.
(슬라이드 22장과 같은 예제입니다.)

```python
high = np.array([18.5, 19.2, 21.0, 19.8, 22.3, 21.5, 23.4])   # 최고 기온
low  = np.array([ 9.2, 10.1, 12.0, 11.4, 13.1, 12.8, 14.0])   # 최저 기온
days = np.array(['월', '화', '수', '목', '금', '토', '일'])
```

```python
일교차 = high - low               # 벡터화: 반복문 없이
일교차
```

```python
np.round(high.mean(), 1), np.round(low.mean(), 1)   # 주간 평균
```

```python
high.max(), days[high.argmax()]    # 최고 기온과 그 요일
```

```python
days[np.argsort(-high)]           # 높은 기온 순서대로 요일
```

> 💡 **감 잡기(선택)**: `high`, `low`의 숫자를 본인 도시의 최근 일주일 기온으로 바꿔서 다시 실행해 보세요. 흐름이 손에 익습니다.
