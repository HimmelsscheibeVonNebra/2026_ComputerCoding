```python
import pandas as pd

# Read input for each day
mon = int(input("Close Mon: "))
tue = int(input("Close Tue: "))
wed = int(input("Close Wed: "))
thu = int(input("Close Thu: "))
fri = int(input("Close Fri: "))

# Build the Series
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
closes = pd.Series([mon, tue, wed, thu, fri], index=days)

# Calculate metrics
mean_val = closes.mean()
high_val = closes.max()
best_day = closes.idxmax()
above_mean_count = (closes > mean_val).sum()

# Print results
print(f"Mean: {mean_val:.1f}")
print(f"High: {high_val}")
print(f"Best day: {best_day} {high_val}")
print(f"Above mean: {above_mean_count}")
```

```python
import pandas as pd

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
typed = ["4200", "4350", "4100", "4400", "4300"]
values = []

#change data of typed (string ) to values (integer/number)
for text in typed:
  values.append(int(text))

closes = pd.Series(values, index=days)

print(closes.mean()) # 4270.0
print(closes.idxmax()) # Thu
```

```python
from google.colab import files
import pandas as pd
import io
import matplotlib.pyplot as plt
import seaborn as sns

# 한글 폰트 설정 (Colab 환경에서 한글 깨짐 방지)
plt.rc('font', family='NanumBarunGothic')

uploaded = files.upload()
file_name = list(uploaded.keys())[0]
df = pd.read_csv(io.BytesIO(uploaded[file_name]))

print("Max Industry Code:", df["Industy_code"].max())

# 간단한 시각화 그래프 생성
plt.figure(figsize=(10, 6))
sns.countplot(data=df, y='Industry', order=df['Industry'].value_counts().index[:10])
plt.title('Top 10 Industries')
plt.xlabel('Count')
plt.ylabel('Industry')
plt.tight_layout()
plt.show()
```
