## 1. 변수와 자료형
변수는 값에 붙이는 **이름표**입니다. `=`는 '같다'가 아니라 **오른쪽 값을 왼쪽 이름에 담는다**(대입)는 뜻입니다.

```python
name = "홍길동"      # str   — 문자열
age = 20             # int   — 정수
height = 173.5       # float — 실수
is_student = True    # bool  — 참/거짓 (True/False는 대문자 시작!)
```

```python
print(name, age, height, is_student)   # 이름표를 부르면 값이 나온다
```

```python
type(name), type(age), type(height), type(is_student)
# (str, int, float, bool) — 자료형 확인은 데이터 분석의 첫 단추
```

```python
age = age + 1   # 오른쪽을 먼저 계산(20+1) → 왼쪽 이름에 다시 담기
age             # 21
```

```python
age += 1        # age = age + 1 의 줄임말
age             # 22
```

> ⚠ 변수 이름은 영문·숫자·밑줄(`_`)만 가능하고 **숫자로 시작할 수 없습니다** — `cool2`는 OK, `2cool`은 오류. 또 `if`, `for` 같은 예약어는 이름으로 쓸 수 없습니다.

## 2. f-string — 문자열에 값 끼워 넣기
따옴표 앞에 `f`를 붙이면 `{ }` 안에 변수를 직접 넣을 수 있습니다. 출력문의 90%는 f-string입니다.

```python
print(f"{name}님은 {age}살입니다.")   # 홍길동님은 22살입니다.
```

```python
pi = 3.14159265
print(f"원주율은 {pi:.2f}")   # 소수점 둘째 자리까지 → 원주율은 3.14
```

```python
print(f"{age}에 8을 더하면 {age + 8}")   # { } 안에서 계산도 가능 → ...30
```

## 3. 산술 연산자
사칙연산에 더해, 데이터 분석에서 단골인 **몫 `//` · 나머지 `%` · 거듭제곱 `**`** 을 확인합니다.

```python
7 + 2, 7 - 2, 7 * 2, 7 / 2   # (9, 5, 14, 3.5)
```

```python
7 // 2    # 몫 → 3
```

```python
7 % 2     # 나머지 → 1 (짝수·홀수 판별의 단골)
```

```python
2 ** 8    # 거듭제곱 → 256
```

```python
365 * 24 * 60   # 1년은 몇 분? → 525600
```

> 💡 `n % 2`는 `n`이 짝수면 0, 홀수면 1 — 잠시 후 '1~100 짝수 합'에서 바로 쓰입니다.

## 4. 비교와 논리
비교의 결과는 `True` / `False`입니다. `and`(그리고) · `or`(또는) · `not`(부정)으로 조건을 조합합니다.

```python
10 > 3, 10 < 3, 10 == 10, 10 != 10   # (True, False, True, False)
```

```python
age >= 20 and age < 30   # True — '20대인가?' (age는 현재 22)
```

```python
not (age >= 20)   # False — 논리 뒤집기
```

## 5. 조건문 — if / elif / else
조건이 참일 때만 실행할 코드는 **들여쓰기(공백 4칸)** 로 구분합니다. 줄 끝의 콜론(`:`)을 잊지 마세요.

```python
score = 85

if score >= 90:
    print("A 학점")      # 조건이 거짓 → 실행되지 않음
else:
    print("90점 미만")   # 이쪽이 실행됨
```

```python
score = 85   # 숫자를 바꿔 가며 여러 번 실행해 보세요

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(f"{score}점 → {grade} 학점")   # 85점 → B 학점
```

> 💡 `elif score >= 80`까지 내려왔다는 건 이미 90 미만이라는 뜻 — 위에서부터 차례로 검사해 **처음 참인 곳에서 멈춥니다**. 그래서 '80 이상 90 미만' 같은 복잡한 조건을 쓰지 않아도 됩니다.

> ⚠ 들여쓰기는 **공백 4칸**으로 통일하고 탭과 섞지 마세요. 들여쓰기가 어긋나면 `IndentationError`, 콜론을 빠뜨리면 `SyntaxError`입니다.

## 6. 반복문 — for + range
`range(1, 10)`은 1부터 9까지 — **끝 숫자는 미포함**입니다. 반복문은 '데이터 개수만큼 뭔가 하기'의 기본 도구입니다.

```python
for i in range(1, 10):            # i가 1, 2, ..., 9로 하나씩 바뀌며 반복
    print(f"3 x {i} = {3 * i}")   # 구구단 3단
```

```python
total = 0                    # 누적합을 담을 그릇 — 반복 전에 0으로 준비

for i in range(1, 101):      # 1부터 100까지
    if i % 2 == 0:           # 짝수라면 (나머지가 0)
        total += i           # 그릇에 더하기

print("1~100 짝수 합:", total)   # 2550
```

```python
list(range(0, 11, 2))   # (시작, 끝(미포함), 간격) → 0, 2, 4, 6, 8, 10
```

> ⚠ `range(1, 10)`에 10은 **포함되지 않습니다** — 1부터 10까지 반복하려면 `range(1, 11)`. 슬라이싱(4주차)·넘파이(5주차)도 같은 규칙입니다.

## 7. while + break
`while`은 조건이 참인 동안 계속 돕니다. '언제 끝날지 모르는' 반복에 쓰고, `break`로 중간에 탈출합니다.

```python
n = 1
while n <= 5:          # 조건(n <= 5)이 참인 동안만 반복
    print(n, "번째 반복")
    n += 1             # 이 줄이 없으면 조건이 영원히 참 → 무한 루프!
```

```python
import random

random.seed(3)         # 시드 고정 → 실행할 때마다 같은 주사위
count = 0

while True:            # 무한 루프를 선언하고
    dice = random.randint(1, 6)
    count += 1
    print(count, "번째 굴림:", dice)
    if dice == 6:      # 6이 나오면
        break          # 루프 탈출
# 2, 5, 5, 2, 3, 5, 4, 6 — 8번째에 6이 나와 종료
```

> ⚠ `while`의 조건을 언젠가 거짓으로 만들지 않으면 프로그램이 영원히 멈춥니다(무한 루프). Colab에서는 셀 왼쪽의 **중단(■) 버튼**으로 정지할 수 있습니다.

## 8. 자주 하는 실수
초보자 4대 오류를 미리 만나 봅시다. 오류 코드는 **주석 처리**해 두었으니, 주석(`# `)을 풀어 실행하면 오류 메시지가 나옵니다 — 오류 메시지를 읽는 연습이 가장 빠른 성장입니다.

```python
name = "홍길동"
age = 20

# print(name + 1)          # TypeError — 문자열과 정수는 +로 못 붙임
print(name + " 나이 " + str(age))   # str()로 감싸 문자열끼리만 붙이기
print(f"{name} 나이 {age}")         # f-string은 자동 변환 — 가장 안전
```

```python
x = 5
# if x = 5:        # SyntaxError — = 는 대입, == 가 비교
if x == 5:
    print("x는 5")   # 실행됨
```

```python
# for i in range(3):
# print(i)         # IndentationError — for 안의 줄은 들여쓰기 필수
for i in range(3):
    print(i)       # 0 / 1 / 2 (한 줄씩 출력)
```

```python
message = "오타 주의"
# print(mesage)    # NameError — mesage라는 변수는 없다 (오타!)
print(message)
```

```python
# 10 / 0           # ZeroDivisionError — 0으로 나눌 수 없음
10 / 0.5           # 0이 아닌 분모라면 OK → 20.0
```

## 9. 종합 연습 — 반복 + 조건 + 누적
오늘 배운 것 전부를 쓰는 클래식 두 문제와 면접 단골 한 문제입니다.

```python
for i in range(1, 6):    # i = 1, 2, 3, 4, 5
    print("*" * i)       # 문자열 * 정수 = 그 횟수만큼 반복 → 별이 자란다
```

```python
total = 0
for i in range(1, 101):
    total += i

print("1부터 100까지 합:", total)   # 5050 — 가우스가 풀었다는 바로 그 문제
```

```python
for i in range(1, 16):
    if i % 15 == 0:      # 3과 5의 공배수(15의 배수)를 먼저 검사!
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

> 💡 **감 잡기(선택)**: 별 찍기에서 `range(1, 6)`을 `range(5, 0, -1)`로 바꾸면 피라미드가 뒤집힙니다 — `range`의 간격은 음수도 됩니다. FizzBuzz에서 `i % 15` 검사를 마지막으로 옮기면 어떻게 될까요?
