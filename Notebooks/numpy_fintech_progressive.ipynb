## 0단계 — 상품 가격을 배열로 · 새 문법: `np.array`, 전체에 곱셈
코드샵에는 상품이 4개 있습니다 — 머그컵 9,000원, 티셔츠 25,000원, 스티커 팩 4,000원, 텀블러 18,000원.
먼저 익숙한 파이썬 리스트에서 시작합니다.

```python
prices_list = [9000, 25000, 4000, 18000]
prices_list * 2     # 리스트에 2를 곱하면?
```

곱셈이 아니라 **반복**이 됩니다. 이제 NumPy 배열로 바꿔 봅니다.

```python
import numpy as np

prices = np.array([9000, 25000, 4000, 18000])   # 머그컵, 티셔츠, 스티커 팩, 텀블러
prices * 2          # 배열에 2를 곱하면?
```

→ `[18000 50000 8000 36000]` — **모든 원소에 2가 곱해집니다.** 리스트와 배열의 이 차이가 오늘 수업의 출발점입니다.

이 `prices` 배열은 마지막 단계까지 계속 사용됩니다.

> ✏️ **해보기**: 곱하는 숫자를 3으로 바꾸고 결과를 먼저 **예상**한 뒤 실행해 보세요.

## 1단계 — 이번 달 판매 수량 · 새 문법: 배열 × 배열 (같은 위치끼리 곱셈)
이번 달에 각 상품이 몇 개 팔렸는지 배열로 만들고, 가격과 곱해 **상품별 매출**을 계산합니다.

```python
quantities = np.array([12, 3, 45, 7])   # 이번 달 판매 수량 — prices와 상품 순서가 같다!
quantities
```

```python
revenue = prices * quantities   # 같은 위치끼리 곱셈
revenue
```

→ `[108000 75000 180000 126000]` — 상품별 매출 (머그컵 108,000원, 티셔츠 75,000원, 스티커 팩 180,000원, 텀블러 126,000원).

⚠ `prices`와 `quantities`의 **상품 순서가 같아야** 정확한 계산이 됩니다. 순서가 다르면 에러는 나지 않지만 **엉뚱한 값**이 나옵니다 — 금융 계산에서 가장 무서운 버그입니다.

> ✏️ **해보기**: 수량 하나를 바꾸고 **어떤 상품의 매출만** 바뀌는지 먼저 예상한 뒤 실행해 보세요.

## 2단계 — 매출 요약 · 새 문법: `.sum()` `.mean()` `.max()` `.min()`
1단계에서 만든 `revenue` 배열을 **요약**해 봅니다. 반복문 없이 배열 전체를 한 번에 집계합니다.

```python
print("총 매출:", revenue.sum(), "원")
print("평균 매출:", revenue.mean(), "원")
```

```python
print("최대 매출:", revenue.max(), "원")
print("최소 매출:", revenue.min(), "원")
```

→ 총 매출 **489,000원**, 평균 **122,250원**, 최대 **180,000원**(스티커 팩), 최소 **75,000원**(티셔츠).

> ✏️ **해보기**: `prices.sum()`, `quantities.mean()`처럼 다른 배열에도 같은 함수를 써 보세요.

## 3단계 — 플래시 세일과 결제 수수료 · 새 문법: 배열 − 배열, `np.round`
이번 달 매출 전체에 **10% 할인**이 적용됐고(0단계의 스칼라 곱셈 재활용!), 결제 대행사가 **수수료 2.5%**를 가져갑니다. 판매자에게 실제로 입금되는 금액을 계산합니다.

```python
discounted = revenue * 0.9   # 10% 할인 → 원래의 90%
discounted
```

```python
fees = discounted * 0.025          # 결제 수수료 2.5%
settlements = discounted - fees     # 배열 − 배열: 실제 입금액
print("수수료:", fees, "원")
print("입금액:", settlements, "원")
```

입금액에 소수(`65812.5`)가 보입니다. 원 단위로 반올림해 봅니다.

```python
np.round(settlements)   # 원 단위 반올림
```

→ `65812.5`가 **65813이 아니라 65812**가 됩니다. 딱 가운데 값은 **가까운 짝수 쪽**으로 반올림되기 때문입니다 (2.5 → 2, 3.5 → 4).
`np.round`는 반올림된 값을 보여줄 뿐 **자료형을 정수로 바꾸는 기능은 아님**도 기억해 두세요.

## 4단계 — 지난달과 합치기 · 새 문법: 배열 + 배열 (같은 위치끼리 덧셈)
지난달 판매 수량 배열을 새로 만들고, 이번 달 수량과 더해 **두 달 누적 판매 수량**을 계산합니다.

```python
last_month = np.array([9, 2, 38, 5])   # 지난달 판매 수량
total_qty = quantities + last_month   # 배열 + 배열
total_qty
```

```python
total_revenue = prices * total_qty   # 1단계에서 배운 배열 × 배열 재사용!
print("누적 매출:", total_revenue, "원")
print("누적 총매출:", total_revenue.sum(), "원")
```

→ 누적 수량 `[21 5 83 12]`, 누적 총매출 **862,000원**. 1단계에서 배운 문법이 새 데이터에 그대로 재사용됩니다 — 점진적으로 자라난다는 게 이 뜻입니다.

> ✏️ **해보기**: 지난달 수량을 바꿔서 누적 총매출이 얼마가 되는지 먼저 계산해 보세요.

## 5단계 — 정산금 예금과 이자 · 새 문법: `np.round(값, 2)` 자릿수 지정
이번 달 입금액 전액을 **연 3% 예금**에 넣고 1년을 기다립니다 (단리, 세금 제외).

```python
cash = settlements.sum()
print("예금 원금:", cash, "원")
```

```python
interest = cash * 0.03                       # 연 3% 단리 1년
print("1년 이자:", np.round(interest, 2), "원")   # 소수점 둘째 자리까지
print("1년 뒤 잔액:", np.round(cash + interest, 2), "원")
```

`np.round(값, 2)`처럼 **자릿수를 지정**하면 소수점 아래 자릿수를 살릴 수 있습니다 — 이자·환율처럼 원 단위보다 작은 돈을 다룰 때 필요합니다.

> ✏️ **해보기**: 금리가 4%라면 이자와 잔액은 얼마가 될까요? `0.03`만 바꿔 다시 실행해 보세요.

## 6단계 — 최종 정산 리포트 · 새 문법: 없음 (배운 것 전부의 조합)
0~5단계에서 만든 변수들로 이번 달 정산 리포트를 출력합니다. **한 줄 한 줄이 이미 배운 문법**이라는 점을 확인하세요.

```python
print("=== 코드샵 이번 달 정산 리포트 ===")
print("상품별 매출 :", revenue, "원")
print("총 매출     :", revenue.sum(), "원")
print("평균 매출   :", revenue.mean(), "원")
print("최대/최소   :", revenue.max(), "/", revenue.min(), "원")
print("할인 후 매출:", discounted.sum(), "원")
print("총 수수료   :", fees.sum(), "원")
print("총 입금액   :", settlements.sum(), "원")
print("1년 뒤 잔액 :", np.round(cash + interest, 2), "원")
```
